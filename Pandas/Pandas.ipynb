```python
import numpy as np
import pandas as pd
```

```python
# Generate random data
data = {
    "ID": np.arange(1, 31),
    "Name": np.random.choice(["Alice", "Bob", "Charlie", "David", "Eve"], size=30),
    "Age": np.random.randint(20, 60, size=30),
    "Salary": np.random.uniform(30000, 120000, size=30).round(2),
    "Department": np.random.choice(["HR", "Finance", "IT", "Marketing"], size=30),
    "Join_Date": pd.date_range(start="2020-01-01", periods=30, freq="ME"),
    "Is_Manager": np.random.choice([True, False], size=30),
    "Performance_Score": np.random.randint(1, 6, size=30),
    "Bonus": np.where(np.random.rand(30) > 0.7, np.random.uniform(500, 5000, size=30).round(2), np.nan)  # Some NaN values
}

# Create DataFrame
df = pd.DataFrame(data)
print(df.shape)
df.head()
```

```python
# How do you group data using groupby()
average_salary = df.groupby("Department")["Salary"].mean()                      # Group by Department and calculate the average salary
grouped_count = df.groupby("Department")["Salary"].count()                      # Count the number of employees in each department
total_salary = df.groupby("Department")["Salary"].sum()                         # Find the total salary per department
aggregations = df.groupby("Department")["Salary"].agg(["mean", "sum", "max"])   # Apply multiple aggregation functions
multiple_sum = df.groupby(["Department", "Is_Manager"])["Salary"].sum()         # Grouping by multiple columns
print("Average Salary:\n", average_salary, '\n')
print("Grouped Count:\n", grouped_count, '\n')
print("Total Salary:\n", total_salary, '\n')
print("Aggregations:\n", aggregations, '\n')
print("Multiple Sum:\n", multiple_sum, '\n')
```
