```python
import numpy as np
```

```python
# Create a tuple with 5 elements
x = (1, 2, 3, 4, 5)
print(x)
print(type(x))
```

```python
# Access the first and last elements of a tuple
x = (1, 2, 3, 4, 5)
first = x[0]
last = x[-1]
print(first)
print(last)
```

```python
# Find the length of a tuple
x = (1, 2, 3)
length = len(x)
print(length)
```

```python
# Count the occurrences of an element in a tuple
x = (1, 2, 3, 2, 4, 2)
count = x.count(2)
print(count)
```

```python
# Find the index of a given element
x = (1, 2, 3, 4, 5)
index = x.index(3)
print(index)
```

```python
# Convert a tuple to a list
x = (1, 2, 3, 4, 5)
x = list(x)
print(x)
print(type(x))
```

```python
# Convert a list to a tuple
x = [1, 2, 3, 4, 5]
x = tuple(x)
print(x)
print(type(x))
```

```python
# Check if an element exists in a tuple
x = (1, 2, 3, 4, 5)
w = x.__contains__(3)
print(w)

if 3 in x:
    print(f"{3} exists in the tuple.")
else:
    print(f"{3} does not exist in the tuple.")
```

```python
# Unpack a tuple into individual variables
x = (1, 2, 3)
a, b, c, = x
print(a, b, c)
```

```python
# Merge two tuples into one
x1 = (1, 2, 3)
x2 = (2, 3, 4, 5, 6)
w1 = x1 + x2
w2 = x1.__add__(x2)
print(w1)
print(w2)
```

```python
# Slice the first 3 elements from a tuple
x = (1, 2, 3, 4, 5)
x = x[:3]
print(x)
```

```python
# Create a tuple with nested tuples
x = ((1, 2), (3, 4), (5, 6))
print(x)
```

```python
# Find the maximum and minimum values in a numeric tuple
x = (1, 2, 3, 4, 5)
print(max(x))
print(min(x))
```

```python
# Create a tuple of even numbers from 1 to 10
x = tuple(range(2, 11, 2))
print(x)
```

```python
# Multiply a tuple by 3
x = (1, 2, 3)
x = x * 3
print(x)
```

```python
# Convert a tuple to a string
x = (1, 2, 3, 4, 5)
x = str(x)
print(x)
print(type(x))
```

```python
# Find the sum of elements in a numeric tuple
x = (1, 2, 3, 4, 5)
print(sum(x))
```

```python
# Check if two tuples are identical
x1 = (1, 2, 3)
x2 = (1, 2, 3)
print(x1 == x2)
```

```python
# Find common elements between two tuples
x1 = (1, 2, 3, 4, 5)
x2 = (4, 5, 6, 7, 8)
x = tuple(set(x1) & set(x2))
print(x)
```

```python
# Swap two tuples
x1 = (1, 2, 3)
x2 = (4, 5, 6)
x1, x2 = x2, x1
print(x1)
print(x2)
```

```python
# Remove an element from a tuple
x = (1, 2, 3, 4, 5)
x = tuple(i for i in x if i != 3)
print(x)
```

```python
# Create a tuple with mixed data types (integers, floats, strings, booleans)
x = (1, 2.0, "Gyan", True)
print(x)
```

```python
# Reverse a tuple without converting it to a list
x = (1, 2, 3, 4, 5)
x = x[::-1]
print(x)
```

```python
# Sort a tuple of numbers in ascending and descending order
x = (1, 4, 2, 5, 3)
x = tuple(sorted(x))
print(x)
x = tuple(sorted(x, reverse=True))
print(x)
```

```python
# Find the most frequent element in a tuple
from collections import Counter
x = (1, 2, 3, 2, 4, 2)
x1 = Counter(x).most_common(1)[0][0]
x2 = max(set(x), key=x.count)
print(x1)
print(x2)
```

```python
# Convert a NumPy array to a tuple
x = np.array([1, 2, 3, 4, 5])
print(x), print(type(x))
y = tuple(x)
print(y), print(type(y))
```

```python
# Generate a tuple of 10 random numbers between 1 and 100
x = tuple(np.random.randint(1, 101, 10))
print(x)
```

```python
# Merge two tuples
x1 = (1, 2, 3, 4, 5)
x2 = (4, 5, 6, 7, 8)
ans = x1 + x2
print(ans)
```

```python
# Merge two tuples and remove duplicate elements
x1 = (1, 2, 3, 4, 5)
x2 = (4, 5, 6, 7, 8)
ans = tuple(set(x1+x2))
print(ans)
```

```python
# Find the intersection of two tuples (common elements).
x1 = (1, 2, 3, 4, 5)
x2 = (4, 5, 6, 7, 8)
ans = tuple(set(x1) & set(x2))
print(ans)
```

```python
# Find the union of two tuples (distinct elements).
x1 = (1, 2, 3, 4, 5)
x2 = (4, 5, 6, 7, 8)
ans = tuple(set(x1) | set(x2))
print(ans)
```

```python
# Find the target elemet in tuple
def func(x, k):
  for i in range(len(x)):
    if x[i] == k:
      print("Yes")
      break

x = (1, 2, 3, 4, 5)
k = 3
func(x, k)
```

```python
# Flatten a tuple of tuples into a single tuple
x = ((1, 2, 3), (4, 5))
x = tuple(sum(x, ()))
print(x)
```
