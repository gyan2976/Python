**20. Consider a (6,7,8) shape array, what is the index (x,y,z) of the 100th element? (★☆☆)**

ravel :
- Purpose: To convert a multidimensional array into a 1D array (flatten it).
- Usage: numpy.ravel(a)
  - Parameters: a is the input array you want to flatten.
  - Returns: A flattened 1D view of the original array (if possible; otherwise, a copy).
  - Example:
    - Create a 2D array
    - array_2d = np.array([[1, 2], [3, 4]])
    - Use ravel to flatten it
    - flattened = np.ravel(array_2d)
    - print(flattened)  # Output: [1 2 3 4]

- unravel_index: Converts a flat index back into coordinates in the original multidimensional array.    

```python
x = np.unravel_index(99, (6, 7, 8))
print(x)
# Explanation : So the 99th element in the flattened version of the array is located at row 1, column 5, and depth 3 in the original 6x7x8 array.
```

**22. Normalize a 5x5 random matrix (★☆☆)**

- Min-Max Normalization rescales values to a range between 0 and 1.
- Z-Score Normalization standardizes values to have a mean of 0 and a standard deviation of 1.

```python
# Min-Max normalization
matrix = np.random.random((5, 5))
min_val = matrix.min()
max_val = matrix.max()
normalized_matrix = (matrix - min_val) / (max_val - min_val)
normalized_matrix
```

```python
# Z-Score Normalization
x = np.random.random((5, 5))
x = (x - np.mean(x) / np.std(x))
x
```

**29. How to round away from zero a float array ? (★☆☆)**

```python
# Author: Charles R Harris

Z = np.random.uniform(-10,+10,10)
print(np.copysign(np.ceil(np.abs(Z)), Z))

# More readable but less efficient
print(np.where(Z>0, np.ceil(Z), np.floor(Z)))
```

**33. How to get the dates of yesterday, today and tomorrow? (★☆☆)**

 - UTC (Coordinated Universal Time)

```python
yesterday = np.datetime64('today') - np.timedelta64(1)
print('Yesterday :', yesterday)
today     = np.datetime64('today')
print('Today :', today)
tomorrow  = np.datetime64('today') + np.timedelta64(1)
print('Tomorrow :', tomorrow)
```

```python
from datetime import datetime

# Get the current date and time
current_time = np.datetime64('now', 's')  # 's' ensures precision up to seconds
print(current_time)
```

```python
from datetime import datetime

# Get the current local date and time
now = datetime.now()

# Format the output to include the date and time with hours, minutes, and seconds
formatted_time = now.strftime("%Y-%m-%d %H:%M:%S")

print(formatted_time)
```
